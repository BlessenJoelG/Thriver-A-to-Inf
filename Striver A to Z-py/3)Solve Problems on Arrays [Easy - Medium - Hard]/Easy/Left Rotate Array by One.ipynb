```python
#optimal+usingSlicing
class Solution:
    def rotateArrayByOne(self, nums):
        temp = nums[1:] + nums[:1]
        nums[:] = temp
```

```python
#optimal2
class Solution:
    def rotateArrayByOne(self, nums):
        first = nums[0]
        for i in range(1,len(nums)):
            nums[i-1] = nums[i]
        nums[-1] = first
```

```python
#better
class Solution:
    def rotateArrayByOne(self, nums):
        temp = [0]*len(nums)
        for i in range(1,len(nums)):
            temp[i-1] = nums[i]
        temp[-1] = nums[0]
        for i in range(len(nums)):
            nums[i] =temp[i]
```
