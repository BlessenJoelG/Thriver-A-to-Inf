```python
#optimalapproach
class Solution:
    def missingNumber(self, nums):
        n = len(nums)
        return ((n*(n+1))//2)-sum(nums)
```

```python
#bruteforce approach
class Solution:
    def missingNumber(self, nums):
        s = set(nums)
        n = len(nums)
        for i in range(n+1):
            if i not in s:
                return i
```
