```python
#bruteforce approach 
class bruteforce:
    def secondLargestElement(self, nums):
        nums.sort()
        largest = nums[len(nums)-1]
        seclargest = -1
        for i in range(len(nums)-1,-1,-1):
            if nums[i]<largest:
                seclargest = nums[i]
                break
        return seclargest
```

```python
#optimal approach
class optimal:
    def secondLargestElement(self, nums):
        largest = nums[0]
        seclargest = -1
        for i in range(1,len(nums)):
            if nums[i]>largest and largest>seclargest:
                largest,seclargest = nums[i],largest
            elif largest>nums[i] and nums[i]>seclargest:
                seclargest = nums[i]
        return seclargest
ans = optimal()
ans.secondLargestElement(nums = [8, 8, 6, 5])
```
