```python
#bruteforce approach
class bruteforce:
    def SortCheck(self,nums):
        if nums == sorted(nums):
            return True
        return False
```

```python
#optimal approach
class optimal:
    def SortCheck(self,nums):
        for i in range(len(nums)-1):
            if nums[i]>nums[i+1]:
                return False
        return True
```
